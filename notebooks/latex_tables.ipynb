# Dataset analysis

## Number of communities over networks and algorithms

```python
import json
import os

# List of datasets (corresponding to *.json files in current directory)
datasets = ["kar", "words", "vote", "pow", "fb-75", "cond-mat", "fb-artist", "dblp"]

# Order of algorithms to appear in the table
algorithms = ["greedy", "louvain", "leiden", "walktrap", "infomap", 
              "label_propagation", "leading_eigenvector", "edge_betweenness", "spinglass"]

# Algorithm short names for the LaTeX header
alg_short = {
    "greedy": "gre",
    "louvain": "louv",
    "leiden": "leid",
    "walktrap": "walk",
    "infomap": "inf",
    "label_propagation": "lab",
    "leading_eigenvector": "eig",
    "edge_betweenness": "btw",
    "spinglass": "spin"
}

# Start building the LaTeX table
latex = r"""\begin{table}[h!]
\centering
\begin{tabular}{lcc""" + "c" * len(algorithms) + "}\n"

# Header row
header = ["\\textbf{Network}", "$|V|$", "$|E|$"] + [f"\\textbf{{{alg_short[a]}}}" for a in algorithms]
latex += " & ".join(header) + r" \\" + "\n\\hline\n"


data_dir = "../outputs_review/dataset_analysis/"
# Process each dataset
for name in datasets:
    filename = f"{name}.json"
    if not os.path.exists(data_dir+filename):
        print(f"Warning: {filename} not found.")
        continue
    
    with open(data_dir+filename, 'r') as f:
        data = json.load(f)
    
    row = [name, str(data["n_nodes"]), str(data["n_edges"])]
    
    for alg in algorithms:
        alg_data = data["detection_algorithms"].get(alg)
        if alg_data:
            if alg_data["Number of Communities"] == "NA":
                row.append("-")
            else:
                row.append(str(alg_data["Number of Communities"]))
        else:
            row.append("-")
    
    latex += " & ".join(row) + r" \\" + "\n"

# End table
latex += r"""\hline
\end{tabular}
\caption{Number of Communities Detected by Different Algorithms}
\end{table}"""

# Output LaTeX code
print(latex)
```

## Budget used for Nabla-CMH

```python
import os
import json
import numpy as np


def extract_budgets_for_latex_table(base_path, tau, beta, alg,seeds):
    datasets = ['kar','words','vote', 'pow', 'fb-75', 'cond-mat']
    latex_rows = []

    
    for dataset in datasets:
        budgets = []
        for seed in seeds:
            file_path = os.path.join(base_path, f'seed_{seed}/{dataset}', f'{alg}', f'tau_{tau}', f'betaFactor_{beta}', 'json_results/nabla-cmh.json')
            with open(file_path, 'r') as file:
                data = json.load(file)
            
            budget = np.mean([data["steps"][i] for i in range(len(data["goal"])) if data["goal"][i] == 1])
            budgets.append(budget)
            total_budget = data["budget"]

        percentage_diff = (np.mean(budgets) / total_budget) * 100
        row_values = [f'${total_budget}$', f'${np.mean(budgets):.1f} ({percentage_diff:.1f}\\%)$']
        
        latex_rows.append(' & '.join(row_values) + ' \\\\')

    return '\n'.join(latex_rows)

tau = 0.5
beta = 2
alg = 'greedy'
base_path = f'../outputs_review/runs/'
seeds=[22,2025,42]
latex_table = extract_budgets_for_latex_table(base_path, tau,beta,alg,seeds)
print(latex_table)
```

## Time comparison between Nabla-CMH and DRL-Agent

```python
import os
import json
import numpy as np


def extract_budgets_for_latex_table(base_path, tau, beta, alg,seeds):
    datasets = ['kar','words','vote', 'pow', 'fb-75', 'cond-mat']
    latex_rows = []

    
    for dataset in datasets:
        times_nabla = []
        times_agent = []
        for seed in seeds:

            file_path_nabla = os.path.join(base_path, f'seed_{seed}/{dataset}', f'{alg}', f'tau_{tau}', f'betaFactor_{beta}', 'json_results/nabla-cmh.json')
            with open(file_path_nabla, 'r') as file:
                data_nabla = json.load(file)
            file_path_agent = os.path.join(base_path, f'seed_{seed}/{dataset}', f'{alg}', f'tau_{tau}', f'betaFactor_{beta}', 'json_results/drl-agent.json')
            with open(file_path_agent, 'r') as file:
                data_agent = json.load(file)
            
            time_nabla = np.mean([data_nabla["time"]])
            time_agent = np.mean([data_agent["time"]])
            times_nabla.append(time_nabla)
            times_agent.append(time_agent)
            
        speedup = np.mean(times_agent) / np.mean(times_nabla)
        row_values = [f'${np.mean(times_nabla):.3f}$', f'${np.mean(times_agent):.3f}$', f'$\\times {speedup:.3f} \\blacktriangle$']
        latex_rows.append(' & '.join(row_values) + ' \\\\')
        

    return '\n'.join(latex_rows)

tau = 0.5
beta = 1
alg = 'greedy'
base_path = f'../outputs_review/runs/'
seeds=[22,2025]
latex_table = extract_budgets_for_latex_table(base_path, tau,beta,alg,seeds)
print(latex_table)
```

## Percentage of additions/deletions for method

```python
import json
import os
import pandas as pd

# List of datasets (corresponding to *.json files in current directory)
datasets = ["kar", "words", "vote", "pow", "fb-75", "cond-mat"]

# Order of algorithms to appear in the table
algorithms = ["random", "betweenness", "degree", "nabla-cmh", "dice", "drl-agent", "roam"]

# Process each dataset
for name in datasets:
    print(f"----- Processing dataset: {name} -----")
    data_dir = f"../outputs_review/runs/seed_22/{name}/greedy/tau_0.5/betaFactor_1/json_results/"
    results = pd.DataFrame(columns=["Algorithm", "Percentage Additions", "Percentage Deletions"])
    for alg in algorithms:

        filename = f"{alg}.json"
        if not os.path.exists(data_dir+filename):
            print(f"Warning: {data_dir+filename} not found.")
            break
    
        with open(data_dir+filename, 'r') as f:
            data = json.load(f)
        
        changes = data["changes"]
        n_additions = 0
        n_deletions = 0
        for dict in changes:
            if alg != "nabla-cmh":
                if len(dict["remove"]) > 0:
                    n_deletions += len(dict["remove"])
                if len(dict["add"]) > 0:
                    n_additions += len(dict["add"])
            else:
                if "removed" in dict and len(dict["removed"]) > 0:
                    n_deletions += len(dict["removed"])
                elif "remove" in dict and len(dict["remove"]) > 0:
                    n_deletions += len(dict["remove"])
                elif "added" in dict and len(dict["added"]) > 0:
                    n_additions += len(dict["added"])
                elif "add" in dict and len(dict["add"]) > 0:    
                    n_additions += len(dict["add"])

        percentage_n_additions = (n_additions / (n_additions+n_deletions)) * 100
        percentage_n_deletions = (n_deletions / (n_additions+n_deletions)) * 100

        # Append the data to the dataframe
        results = results.append({
            "Algorithm": alg,
            "Percentage Additions": percentage_n_additions,
            "Percentage Deletions": percentage_n_deletions
        }, ignore_index=True)

    print(results)
    
```

## Parameter sensitivity (SR)

```python
import numpy as np
import os
import json


def _fmt_sr_cell(mean, std, rank):
    """Format a single SR cell: rank 0 -> bold (best), rank 1 -> underline (second)."""
    s = f"{mean:.2f} \\pm {std:.2f}"
    if rank == 0:
        return f"$\\mathbf{{{s}}}$"
    elif rank == 1:
        return f"$\\underline{{{s}}}$"
    else:
        return f"${s}$"


def build_sr_param_sensitivity_table(base_path, taus, betas, methods, dataset, alg, seeds):
    beta_labels = {0.5: r"$\mu/2$", 1: r"$\mu$", 2: r"$2\mu$"}
    method_headers = {
        "NABLA": r"\method{}\,(ours)",
        "DRL": "DRL-Agent",
        "DICE": "DICE",
        "RAND": "Random",
    }

    body_rows = []
    for tau in taus:
        body_rows.append(r"        \midrule")
        body_rows.append(r"        \multirow{%d}{*}{%s} " % (len(betas), tau))
        for beta in betas:
            means, stds = [], []
            for method in methods:
                srs = []
                for seed in seeds:
                    file_path = os.path.join(
                        base_path, f'seed_{seed}', dataset, alg,
                        f'tau_{tau}', f'betaFactor_{beta}', 'plots/metrics.json'
                    )
                    with open(file_path, 'r') as f:
                        data = json.load(f)
                    # SR == "goal" (success rate), kept on the 0-100 percentage scale
                    srs.append(data["goal"][method])
                means.append(float(np.mean(srs)))
                stds.append(float(np.std(srs)))

            # Rank on the displayed (2-decimal) values so visual ties share a style.
            rmeans = [round(m, 2) for m in means]
            best = max(rmeans)
            rest = [v for v in rmeans if v < best]
            second = max(rest) if rest else None
            ranks = []
            for v in rmeans:
                if v == best:
                    ranks.append(0)            # best -> bold
                elif second is not None and v == second:
                    ranks.append(1)            # second best -> underline
                else:
                    ranks.append(2)

            cells = [_fmt_sr_cell(means[i], stds[i], ranks[i]) for i in range(len(methods))]
            body_rows.append(f"        & {beta_labels[beta]} & " + " & ".join(cells) + r" \\")

    header = " & ".join(method_headers[m] for m in methods)
    latex = r"""\begin{table}[H]
    \centering
    \caption{Impact of $\tau$ and $\beta$ on the success rate (SR) in the symmetric setting (\,$\f$: \textit{greedy}) on the \texttt{%s} dataset for \method{}, DRL-Agent, DICE, and Random. Bold indicates the best (highest) value; underlined values are second best.}
    \label{tab:param_sens_sr}
    \scalebox{0.8}{
    \begin{tabular}{ccccccccccc}
        \toprule
        \multirow{2}{*}{$\tau$} & \multirow{2}{*}{$\beta$} & \multicolumn{4}{c}{Algorithm} \\
        \cmidrule(lr){3-6}
        & & %s  \\
%s
        \bottomrule
    \end{tabular}
    }
\end{table}""" % (dataset, header, "\n".join(body_rows))
    return latex


taus = [0.3, 0.5, 0.8]
betas = [0.5, 1, 2]
methods = ["NABLA", "DRL", "DICE", "RAND"]
dataset = 'vote'
alg = 'greedy'
seeds = [22, 2025, 42]
base_path = f'../outputs_review/runs/'
latex_table = build_sr_param_sensitivity_table(base_path, taus, betas, methods, dataset, alg, seeds)
print(latex_table)
```

## Significance of impact

```python
import numpy as np
import enum
import os
import json

class DatasetNames(enum.Enum):
    """Enum class for the dataset names"""

    KAR = "kar"
    WORDS = "words"
    VOTE = "vote"
    POW = "pow"
    FB_75 = "fb-75"
    COND_MAT = "cond-mat"

class EvasionAlgorithmsNames(enum.Enum):
    RAND = "random"
    DEG = "degree"
    BETW = "betweenness"
    ROAM = "roam"
    DICE = "dice"
    NABLA = "nabla-cmh"
    DRL = "drl-agent"
    GRE = "greedy"

def exp_to_latex(sci_str, precision=1):
    """Converte una stringa in notazione scientifica in formato LaTeX."""
    num = float(sci_str)
    base_str = f"{num:.{precision}e}"  # es. '1.23e-04'
    base, exp = base_str.split('e')
    return f"{base} \\cdot 10^{{{int(exp)}}}"

def extract_values_for_latex_modification_importance(base_path):
    latex_rows = []
    for dataset in datasets:
        row_values = [f'{getattr(DatasetNames, dataset).value}' if dataset!='cond-mat' else 'arxiv']
        for method in evasion_algs:
            file_path = os.path.join(base_path, 'total_impact.json')
            with open(file_path, 'r') as file:
                data = json.load(file)
            avg_pr_mean = data[f'{getattr(DatasetNames,dataset).value}'][method]["mean"]
            avg_pr_std = data[f'{getattr(DatasetNames,dataset).value}'][method]["std"]
            
            row_values.append(f"${exp_to_latex(avg_pr_mean)} \\pm {exp_to_latex(avg_pr_std)}$")
        latex_rows.append(' & '.join(row_values) + ' \\\\')

    return '\n'.join(latex_rows)


evasion_algs = ["NABLA", "DICE", "BETW"] 
datasets = ["KAR", "WORDS", "VOTE", "POW", "FB_75", "COND_MAT"]
base_path = f'../outputs_review/all_datasets/impact_significance'
latex_table = extract_values_for_latex_modification_importance(base_path)
print(latex_table)
```
